```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=10000,n_features=11,n_classes=2,random_state=12)
```

```python
print(x.shape)
print(y.shape)
```

```python
print(type(x))
```

```python
pd.DataFrame(x)
```

```python
pd.DataFrame(y)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=12
)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
plt.scatter(y_pred,y_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy=accuracy_score(y_pred,y_test)
print(accuracy)
print(confusion_matrix(y_test,y_pred))
report=classification_report(y_test,y_pred)
print(report)
```

## HyperParameter Tuning  the model 


```python
from sklearn.model_selection import GridSearchCV

model=LogisticRegression()
C=[0.1,0.01,1.0,10,100]
penalty=['l1', 'l2', 'elasticnet']
solver=['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']

```

```python
params = [
    {
        'solver': ['lbfgs'],
        'penalty': ['l2'],
        'C': [100, 10, 1, 0.1, 0.01]
    },
    {
        'solver': ['newton-cg'],
        'penalty': ['l2'],
        'C': [100, 10, 1, 0.1, 0.01]
    },
    {
        'solver': ['liblinear'],
        'penalty': ['l1', 'l2'],
        'C': [100, 10, 1, 0.1, 0.01]
    },
    {
        'solver': ['sag'],
        'penalty': ['l2'],
        'C': [100, 10, 1, 0.1, 0.01]
    },
    {
        'solver': ['saga'],
        'penalty': ['l1', 'l2'],
        'C': [100, 10, 1, 0.1, 0.01]
    }
]
```

```python
from sklearn.model_selection import StratifiedKFold
new_cv=StratifiedKFold()
```

```python
newmodel=GridSearchCV(param_grid=params,scoring='accuracy',cv=new_cv,n_jobs=-1,estimator=model)
```

```python
newmodel.fit(x_train,y_train)
```

```python
newmodel.best_estimator_
```

```python
newmodel.cv_results_['rank_test_score']
```

```python
newmodel.cv_results_['params']
```

```python
newmodel.cv_results_['mean_test_score']
```

```python
newmodel.best_score_
```

```python
y_pred_new=newmodel.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy=accuracy_score(y_pred_new,y_test)
print(accuracy)
print(confusion_matrix(y_pred_new,y_test))
print(classification_report(y_pred_new,y_test))
```

```python
from sklearn.model_selection import RandomizedSearchCV

model=LogisticRegression()
C=[0.1,0.01,1.0,10,100]
penalty=['l1', 'l2', 'elasticnet']
solver=['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
```

```python
paramss=dict(penalty=penalty,C=C,solver=solver)
```

```python
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
model_randomized=RandomizedSearchCV(estimator=model,n_iter=10,cv=cv,param_distributions=paramss)
```

```python
model_randomized.fit(x_train,y_train)
```

```python
model_randomized.best_estimator_
```

```python
model_randomized.best_score_
```

```python
y_pred_latest=model_randomized.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy_random=accuracy_score(y_pred_latest,y_test)
print(accuracy_random)
print(confusion_matrix(y_pred_latest,y_test))
print(classification_report(y_pred_latest,y_test))
```

```python
import pickle
pickle.dump(newmodel,open('final_model.pkl','wb'))
pickle.dump(model_randomized,open('latest_model.pkl','wb'))
```
